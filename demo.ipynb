## View Results

Results are saved to `results/<run_id>/`:
- `summary.json`: Aggregated results with statistics
- `summary.csv`: Detailed per-iteration results
- `system.json`: System configuration


```python
# View summary from most recent run
import json
from pathlib import Path

# Get most recent result directory
result_dirs = sorted(Path("results").glob("*/"), key=lambda p: p.stat().st_mtime, reverse=True)
if result_dirs:
    latest_dir = result_dirs[0]
    summary_path = latest_dir / "summary.json"
    
    if summary_path.exists():
        with open(summary_path) as f:
            data = json.load(f)
        
        print(f"Results from: {latest_dir.name}")
        print("\n=== Summary ===")
        print(json.dumps(data["summary"], indent=2))
    else:
        print(f"No summary.json found in {latest_dir}")
else:
    print("No results found")
```
